```python
#Rede neural perceptron de multcamada
#importando bibliotecas
import numpy as np
```

```python
#Criando a função de sigmoid
def sigmoid(soma):
    return 1/(1 + np.exp(-soma))
```

```python
#Criando a derivada da função sigmoid
def sigmoidDerivada(sig):
    return sig * (1 - sig)
```

```python
a = sigmoid(0.5)
print(a)
```

```python
b = sigmoidDerivada(a)
print(b)
```

```python
#Entrada das informações da tabala verdade
entradas = np.array([[0,0],
                     [0,1],
                     [1,0],
                     [1,1]])
```

```python
#Valores desejados
saidas = np.array([[0],[1],[1],[0]])
```

```python
#Pesos(sinapses) para aprender os padrões da camada de entrada para camada escondida
pesos0 = np.array([[-0.424, -0.740, -0.961],
                    [0.358, -0.577, -0.469]])
print(f'Pesos da camada de entrada para a camada escondida {pesos0}')
```

```python
#Pesos(sinapses) para aprender os padrões da camada escondida para camada de saída
pesos1 = np.array([[-0.017],[-0.893],[0.148]])
print(f'Pesos da camada escondida para a camada de saída {pesos1}')
```

```python
#Criar a quantidade de vezes que pode fazer o looping
epocas = 100
```

```python
for j in range(epocas):
    camadaEntrada = entradas

    somaSinapse0 = np.dot(camadaEntrada, pesos0) #Pesos
    camadaOculta = sigmoid(somaSinapse0)

    somaSinapse1 = np.dot(camadaOculta, pesos1) #Pesos
    camadaSaida = sigmoid(somaSinapse1)

    erroCamadaSaida = saidas - camadaSaida
    mediaAbsoluta = np.mean(np.abs(erroCamadaSaida)) #Media absoluta do Erro

    derivadaSaida = sigmoidDerivada(camadaSaida)
    deltaSaida = erroCamadaSaida * derivadaSaida

    pesos1Transposta = pesos1.T
    deltaSaidaXPeso = deltaSaida.dot(pesos1Transposta)
    deltaCamadaOculta = deltaSaidaXPeso * sigmoidDerivada(camadaOculta)
```

```python
print(f'Resultado camada de entrada para camada oculta {somaSinapse0}')
```

```python
print(f'Resultado da função de ativação sigmoid para a camada oculta {camadaOculta}')
```

```python
print(f'Resultado camada de oculta para camada de saída {somaSinapse1}')
```

```python
print(f'Resultado da função de ativação sigmoid para a camada de saida {camadaSaida}')
```

```python
print(f'Erros da camada de saida {erroCamadaSaida}')
```

```python
print(f'Media a absoluta do erro {mediaAbsoluta}')
```

```python
print(f'Derivada do resultado da ativação da camada de saida {derivadaSaida}')
```

```python
print(f'Delta camada de saida {deltaSaida}')
```

```python
print(f'Delta camada de saida multiplicada pelo peso {deltaSaidaXPeso}')
```

```python
print(f'Delta camada oculta {deltaCamadaOculta}')
```
